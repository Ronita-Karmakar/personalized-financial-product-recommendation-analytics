```python
import pandas as pd
import numpy as np
import random

from faker import Faker
from datetime import datetime, timedelta
```

```python
fake = Faker("en_IN")
```

```python
random.seed(42)
np.random.seed(42)
Faker.seed(42)
```

```python
NUM_CUSTOMERS = 10000

occupation_income = {
    "Student": (0, 200000),
    "Teacher": (300000, 800000),
    "Software Engineer": (600000, 1800000),
    "Data Analyst": (500000, 1500000),
    "Doctor": (1000000, 3000000),
    "Business Owner": (800000, 5000000),
    "Government Employee": (400000, 1200000),
    "Accountant": (400000, 1000000),
    "Sales Executive": (300000, 900000),
    "Retired": (150000, 600000)
}

cities = {
    "Kolkata": "West Bengal",
    "Delhi": "Delhi",
    "Mumbai": "Maharashtra",
    "Bengaluru": "Karnataka",
    "Hyderabad": "Telangana",
    "Chennai": "Tamil Nadu",
    "Pune": "Maharashtra",
    "Ahmedabad": "Gujarat",
    "Jaipur": "Rajasthan",
    "Lucknow": "Uttar Pradesh"
}

def get_customer_segment(income):
    
    if income < 500000:
        return "Retail"
    
    elif income <= 1500000:
        return "Premium"
    
    else:
        return "VIP"
```

```python
print(get_customer_segment(250000))
print(get_customer_segment(850000))
print(get_customer_segment(2200000))
```

```python
customer_ids = [
    f"C{i:05d}"
    for i in range(1, NUM_CUSTOMERS + 1)
]

customer_ids[:5]
```

```python
customers = []

for customer_id in customer_ids:

    # Generate Name
    first_name = fake.first_name()
    last_name = fake.last_name()

    # Gender
    gender = random.choices(
        ["Male", "Female", "Other"],
        weights=[49, 49, 2],
        k=1
    )[0]

    # Age between 21 and 70
    age = random.randint(21, 70)

    dob = (
        datetime.today()
        - timedelta(days=age * 365 + random.randint(0, 364))
    ).date()

    # Occupation
    occupation = random.choice(list(occupation_income.keys()))

    # Income based on occupation
    income = random.randint(
        occupation_income[occupation][0],
        occupation_income[occupation][1]
    )

    # Customer Segment
    segment = get_customer_segment(income)

    # City & State
    city = random.choice(list(cities.keys()))
    state = cities[city]

    # Marital Status
    marital_status = random.choices(
        ["Single", "Married", "Divorced", "Widowed"],
        weights=[45, 45, 5, 5],
        k=1
    )[0]

    # Email
    email = (
        first_name.lower()
        + "."
        + last_name.lower()
        + str(random.randint(1,999))
        + "@email.com"
    )

    # Phone
    phone = str(random.randint(6000000000, 9999999999))

    # Join Date
    join_date = fake.date_between(
        start_date="-8y",
        end_date="today"
    )

    # Store Record
    customers.append({
        "Customer_ID": customer_id,
        "First_Name": first_name,
        "Last_Name": last_name,
        "Gender": gender,
        "DOB": dob,
        "Email": email,
        "Phone": phone,
        "City": city,
        "State": state,
        "Occupation": occupation,
        "Annual_Income": income,
        "Marital_Status": marital_status,
        "Customer_Segment": segment,
        "Join_Date": join_date
    })
```

```python
customers_df = pd.DataFrame(customers)
```
